```python
def favorite_things(name, movie, music):
    return f"My name is {name}.\nMy favorite movie is {movie}.\nMy favorite musician or band is {music}."

print(favorite_things("Ajani", "Spider-Man", "Brent Faiyaz"))
```

```python
def why_im_here(purpose):
    return purpose

print(why_im_here("I joined the Year Up United Data Analyst program to develop my data analysis skills and prepare for a career in the technology and data field."))
```

```python
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."
```

```python
print(favorite_place("New York City", "there is always something interesting to do"))
```

```python
print(favorite_place("The beach", "it is peaceful and relaxing"))
```
